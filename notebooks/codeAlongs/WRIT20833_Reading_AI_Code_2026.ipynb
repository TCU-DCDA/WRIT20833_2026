# Setup

````python
# SETUP — run this cell first. It's plumbing: the same tools as the Term Frequency code-along.
import re, os, urllib.request
from collections import Counter

def split_into_words(any_chunk_of_text):
    lowercase_text = any_chunk_of_text.lower()
    # Split on any run of non-letter/number characters (spaces, punctuation, newlines).
    return re.split(r"\W+", lowercase_text)

stopwords = ["i", "me", "my", "myself", "we", "our", "ours", "ourselves", "you", "your", "yours", "yourself", "yourselves", "he", "him", "his", "himself", "she", "her", "hers", "herself", "it", "its", "itself", "they", "them", "their", "theirs", "themselves", "what", "which", "who", "whom", "this", "that", "these", "those", "am", "is", "are", "was", "were", "be", "been", "being", "have", "has", "had", "having", "do", "does", "did", "doing", "a", "an", "the", "and", "but", "if", "or", "because", "as", "until", "while", "of", "at", "by", "for", "with", "about", "against", "between", "into", "through", "during", "before", "after", "above", "below", "to", "from", "up", "down", "in", "out", "on", "off", "over", "under", "again", "further", "then", "once", "here", "there", "when", "where", "why", "how", "all", "any", "both", "each", "few", "more", "most", "other", "some", "such", "no", "nor", "not", "only", "own", "same", "so", "than", "too", "very", "s", "t", "can", "will", "just", "don", "should", "now", "ve", "ll", "amp"]

def top_meaningful_words(text, n):
    meaningful_words = []
    for word in split_into_words(text):
        if word != "" and word not in stopwords:
            meaningful_words.append(word)
    return Counter(meaningful_words).most_common(n)

def load_text(filename):
    for path in (filename, os.path.join("notebooks", "data", filename),
                 os.path.join("data", filename), os.path.join("..", "data", filename)):
        if os.path.exists(path):
            return open(path, encoding="utf-8").read()
    url = "https://raw.githubusercontent.com/TCU-DCDA/WRIT20833_2026/main/notebooks/data/" + filename
    return urllib.request.urlopen(url).read().decode("utf-8")

# The 93 YouTube comments on the 2025 Texas Ten Commandments law, from HW1 and HW2:
comments_text = load_text("ten_commandments/tc_youtube_comments.txt")
print("Setup complete.")
````

# Part 1 — Read before you run *(checklist 1 and 2)*

Someone asked an AI: *"Write a Python function that counts how often each word appears in a text."*
This is what came back. It's the function from the lecture. Run the cell: it only **defines** the
function, so nothing prints yet.

````python
def count_words(text):
    counts = {}
    for word in text.split():
        counts[word] = counts.get(word, 0) + 1
    return counts
````

You've written this function before. Here's the version you built by hand in the Dictionaries &
Functions code-along. It has a new name, `count_words_by_hand`, so it doesn't replace the AI's
`count_words`: two functions can't share a name, and the one you define last wins. Run it to define it.

````python
# the version you built by hand in Dictionaries & Functions
def count_words_by_hand(word_list):
    counts = {}
    for word in word_list:
        if word in counts:
            counts[word] = counts[word] + 1
        else:
            counts[word] = 1
    return counts
````

Now run both on the same sentence. Look at what each one is **given** inside the parentheses.

### The one new piece: `.get`

You glimpsed `.get` once, in the Strings code-along's Sneak Preview, but nobody explained it. That's
exactly the situation borrowed code puts you in. **Find out before you trust it.**

````python
tally = {"god": 2}

# predict first: what will each line print?
# my guess:
print(tally.get("god", 0))
print(tally.get("texas", 0))
print(tally)       # did .get add "texas" to the dictionary?
````

The structure:
```python
dictionary.get(key, default)    # the key's value if the key is there; otherwise the default
```
`.get` only *looks*. It never changes the dictionary. So `counts.get(word, 0) + 1` means "the count so far,
or 0 if this is the first time, plus one," and the `=` stores the new count. That's your `if`/`else`, in one line.

# Part 4 — The helpful extra *(all five)*

Next request: *"Write a function that returns the 10 most meaningful words in a text, skipping
stopwords."* The AI wrote this. It uses one-letter variable names, the way AI (and many coders) write: `w` for
each word. Short names are a habit, not a rule. Run it to define it.

````python
def top_words(text, n=10):
    words = []
    for w in text.lower().split():
        if w not in stopwords and len(w) > 3:
            words.append(w)
    return Counter(words).most_common(n)
````

One new piece of syntax: **`n=10`** gives `n` a **default**. Call `top_words(comments_text)` and
`n` is 10; call `top_words(comments_text, 5)` and it's 5.

# 4. Write a Small Function

You just wrote a keyword test inside a loop. Wrapping it in a function gives it a name, `mentions`,
so you can reuse it with any post and any word, as you did with `is_long` in the Dictionaries &
Functions code-along.

- It takes **two parameters**: the post to check, and the word to look for
- Its `return` line hands back a comparison, so the answer is `True` or `False`, like `is_recent`
  in that code-along

````python
def mentions(post, word):
    # True if the post's text contains `word`, in any capitalization
    return word.lower() in post["text"].lower()

print(mentions(POSTS[0], "quiet"))   # P1
print(mentions(POSTS[3], "quiet"))   # P4
````

# 6. Fill a Gap: Any Word From a List

HW1 Part C needs this pattern. It's the same one as "A loop inside a loop" in the Lists & Loops
code-along, here on longer text.

Step 5 asked whether to change the rule. A common change is to check a **list** of words instead of
one: does this post mention *any* of them? That takes a loop inside a loop. The outer loop visits
each post, and the inner loop checks that post against each word.

````python
# predict first — there are 8 posts. How many codes will sound_posts hold?
# My guess:
sound_words = ["quiet", "silence", "noise", "talk"]

sound_posts = []
for post in POSTS:
    for word in sound_words:
        if word in post["text"].lower():
            sound_posts.append(post["code"])

print(len(sound_posts), sound_posts)
````

Eleven, from eight posts. P2 mentions both *quiet* and *talk*, so it went in twice, and so did
P4, P5, and P6. The `.append()` runs once per **matching word**, and we wanted once per **post**.

The fix is a **True/False switch**: a variable that remembers whether this post has matched anything
yet.

The structure:
```python
new_list = []
for item in my_list:
    found = False
    for thing in other_list:
        if thing in item:
            found = True
    if found:
        new_list.append(item)
```

- `found` starts at `False` for **every** item, so it's set inside the outer loop, before the inner one
- The inner loop only flips the switch to `True`. It doesn't append anything
- The `.append()` sits in the **outer** loop, after the inner loop has finished, so each item is
  added once
- Indentation tells Python which loop each line belongs to, so line it up carefully

````python
sound_posts = []
for post in POSTS:
    found = False                      # no match yet for this post
    for word in sound_words:
        if word in post["text"].lower():
            found = True               # at least one word matched
    if found:                          # back in the outer loop: add the post once
        sound_posts.append(post["code"])

print(len(sound_posts), sound_posts)
````

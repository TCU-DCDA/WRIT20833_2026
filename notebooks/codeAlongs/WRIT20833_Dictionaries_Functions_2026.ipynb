# Dictionaries

Last time, **lists** stored items in order. But a lot of cultural data isn't just a sequence — it's
**labeled**. A song has a title, an artist, a year. A **dictionary** stores **key–value pairs**: each
value gets a name (its *key*) that you look it up by.

The structure:
```python
my_dict = {
    "key1": value1,
    "key2": value2,
}

my_dict["key1"]    # look up a value by its key → value1
```

- Created with curly braces `{}`
- Each entry is `key: value`, entries separated by commas
- Keys are usually strings, in quotes; values can be any type (strings, numbers, even lists)
- Each key appears only once in a dictionary
- Look a value up with square brackets and the key: `my_dict["key"]`. A list uses a *position*
  (`my_list[0]`); a dictionary uses a *name*

````python
album = {
    "title": "Renaissance",
    "artist": "Beyoncé",
    "year": 2022,
    "tracks": 16
}

print(album["title"])
print(album["artist"], "—", album["year"])
````

### Keys and values
You can ask a dictionary for all of its keys or all of its values.

The structure:
```python
my_dict.keys()      # every key
my_dict.values()    # every value
```

- These are **methods**, like `.split()` on a string: a dot, the method name, then parentheses

### Adding and changing entries
Dictionaries are editable. Assign to a key to **add** it (if it's new) or **change** it (if it
already exists).

The structure:
```python
my_dict["key"] = value
```

- Same `=` you use for variables, but the left side names a key inside the dictionary
- If the key is new, Python **adds** it; if the key already exists, the new value **replaces** the old one
- Python won't warn you which one happened, so check your spelling of the key

For example:
```python
album["label"] = "Parkwood"                # new key → added
album["artist"] = "Beyoncé Knowles-Carter" # existing key → changed
```


# Counting with a Dictionary

Here's the move that powers most text analysis: use a dictionary to **count** things. The **key** is
the thing; the **value** is how many times you've seen it. This is exactly how we'll count words in
real comments next week.

One new piece makes it work:
```python
key in my_dict    # True if key is already one of the dictionary's keys, otherwise False
```

- `in` checks the **keys**, not the values
- The counting pattern combines this check with adding and changing entries: if the thing is new,
  **add** it with a count of 1; if it's already there, **change** its count to one more

````python
# count how often each word appears
words = ["god", "law", "god", "freedom", "god", "law"]

counts = {}                 # start with an empty dictionary
for word in words:
    if word in counts:
        counts[word] = counts[word] + 1   # seen before — add one
    else:
        counts[word] = 1                  # first time — start at one

print(counts)
````

# Functions

A **function** packages code into a reusable tool. You *define* it once with `def`, give it inputs
(*parameters*), and it can hand back a result with `return`.

The structure:
```python
def function_name(parameter):
    do something with parameter
    return result

function_name(value)    # call it: value goes in as parameter, result comes back
```

- `def` starts the definition; the name follows the same rules as variable names
- The **parameter** is a placeholder variable for whatever input you give the function
- The line ends with a colon, and the body is indented, as in loops and `if` statements
- `return` hands a value back and ends the function
- `return` is not the same as `print`. `print` only shows a value on the screen; `return` gives it back
  to the code that called the function, so you can print it, save it to a variable, or use it in a loop.
  That's why the examples below wrap each call in `print()`
- Defining a function doesn't run it. It runs each time you **call** it by name with parentheses

````python
def greet(name):
    return f"Welcome, {name}!"

print(greet("Frog fan"))
print(greet("Dr. Rode"))
````

### Why functions? Write the recipe once.
Without a function you'd copy-paste the same lines every time you needed them. With one, you write the
logic **once** and reuse it — and if you fix it, you fix it everywhere at once.

````python
def classify_year(year):
    if year <= 1964:
        return "Boomer"
    elif year <= 1980:
        return "Gen X"
    elif year <= 1996:
        return "Millennial"
    else:
        return "Gen Z"

print(classify_year(1990))
print(classify_year(2005))
````

The payoff: call your function inside a loop to process a whole list.

The structure:
```python
for item in my_list:
    print(function_name(item))    # call the function once per item
```

### More than one input
Functions can take several parameters.

The structure:
```python
def function_name(parameter1, parameter2):
    return something using parameter1 and parameter2

function_name(value1, value2)
```

- Separate parameters with commas, both when you define the function and when you call it
- Values match parameters **by position**: the first value goes to the first parameter, and so on

# Putting It All Together

Now combine **functions** and **dictionaries** into one reusable tool: a function that takes a list of
words and returns a dictionary of how often each appears. This is a first draft of the term-frequency
counter you'll use on real data next week.

````python
def count_words(word_list):
    counts = {}
    for word in word_list:
        if word in counts:
            counts[word] = counts[word] + 1
        else:
            counts[word] = 1
    return counts

sample = ["god", "law", "freedom", "god", "school", "god", "law"]
print(count_words(sample))
````

# Sneak Preview: Where This Is Going

Next week we leave practice data behind for **real** text — 93 actual YouTube comments about a Texas
law. Your `count_words` function is the seed of **term frequency**: finding what a text is *really*
about by counting its words. Python even ships with a tool, `Counter`, that does what you just wrote.

The structure:
```python
from collections import Counter     # load a tool that comes with Python

Counter(my_list)                    # counts each item, like count_words
Counter(my_list).most_common(n)     # the n most frequent items, as (item, count) pairs
```

````python
from collections import Counter

comment = "god god law freedom god school law god"
words = comment.split()

print(Counter(words).most_common(3))   # the 3 most frequent words
````

That little count is already a tiny act of *interpretation*: **we** chose to split on spaces,
**we** decided every word counts equally. Next week we'll ask what choices like these reveal — and what
they flatten.

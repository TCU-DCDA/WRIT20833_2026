### `if` / `elif` / `else`

An `if` statement runs code only when a condition is `True`. Add `elif` (else if) for additional checks, and `else` as a catch-all.

The structure:
```python
if condition:
    do this
elif another_condition:
    do this instead
else:
    do this if nothing above was True
```

- Each condition line ends with a **colon** `:`, and the code underneath is **indented**
- Python checks from the top and runs only the **first** block whose condition is `True`
- `elif` and `else` are optional; you can have as many `elif`s as you need

````python
tcu_score = 34
smu_score = 31

if tcu_score > smu_score:
    print("TCU wins!")
elif tcu_score == smu_score:
    print("Headed to overtime")
else:
    print("Better luck next year")
````

Two conditions can be joined with `and` (both must be true) or `or` (either one will do).

The structure:
```python
if condition1 and condition2:    # True only if both are True
if condition1 or condition2:     # True if either one is True
if not condition:                # not flips it: True becomes False, False becomes True
```

A condition is just a `True`/`False` value, so you can also store it in a variable and use the name later:

```python
is_adult = age >= 18             # is_adult now holds True or False
if is_adult and not is_banned:
    ...
```

# For Loops

A **for loop** repeats an action for every item in a list. This is where coding gets powerful — instead of writing the same code over and over, you write it once and the loop handles the rest.

The structure:
```python
for item in my_list:
    do something with item
```

- `item` is a temporary variable that takes on each value in the list, one at a time
- The indented code runs once per item
- You can name `item` whatever makes sense (`name`, `tag`, `year`, etc.)

````python
# loop through our platforms list
for platform in platforms:
    print(f"Checking: {platform}")
````

### Loops + Conditionals

The real power shows up when you combine loops with `if` statements — now you can process a list *and* make decisions about each item.

The structure:
```python
for item in my_list:
    if condition about item:
        do this
    else:
        do this instead
```

- Two levels of indentation: the `if` is inside the loop, and its code is inside the `if`
- The whole `if`/`else` runs once for every item

````python
# filter: which platforms have short names?
for platform in platforms:
    if len(platform) <= 5:
        print(f"{platform} — short name ({len(platform)} chars)")
    else:
        print(f"{platform} — longer name ({len(platform)} chars)")
````

### Counting with loops

A common pattern: start a counter at 0, then add to it inside a loop.

The structure:
```python
count = 0
for item in my_list:
    if condition about item:
        count = count + 1
```

- The counter starts **before** the loop. Inside the loop it would reset to 0 on every item
- `count = count + 1` means "take the current count, add one, store it back"

````python
# count how many hashtags mention TCU
tags = ["#TCU", "#GoFrogs", "#FortWorth", "#TCU", "#BigXII", "#TCU"]

tcu_count = 0

for tag in tags:
    if tag == "#TCU":
        tcu_count = tcu_count + 1

print(f"TCU was mentioned {tcu_count} times out of {len(tags)} total tags")
````

You can keep **more than one counter** at once: one for each category, all started at 0 before the
loop. Inside the loop, an `if` / `elif` / `else` decides which counter gets the `+ 1`.

### Building new lists from old ones

Another essential pattern: start with an empty list, loop through a source list, and `.append()` items that meet your criteria.

This is like going through a library catalog and pulling only the books that match your research topic.

The structure:
```python
new_list = []
for item in my_list:
    if condition about item:
        new_list.append(item)
```

- The empty list starts **before** the loop, like a counter
- To keep the items that *don't* match, test with `not in`: `"#" not in tag` is `True` for a tag
  with no `#`

````python
# pull out only the short platform names
short_names = []

for platform in platforms:
    if len(platform) <= 5:
        short_names.append(platform)

print("Short platform names:", short_names)
````

### Using `enumerate()` — tracking position

Sometimes you need to know both the item *and* its position. `enumerate()` gives you both.

The structure:
```python
for position, item in enumerate(my_list, 1):
    do something with position and item
```

- `enumerate()` hands the loop a **pair** each time: a position number and an item
- So the loop gets **two** variables, separated by a comma: the position first, then the item
- The `1` says to start counting at 1. Leave it out and counting starts at 0

````python
# create a numbered playlist
playlist = ["Not Like Us", "Espresso", "Birds of a Feather", "APT.", "Die With a Smile"]

for position, song in enumerate(playlist, 1):  # start counting at 1
    print(f"{position}. {song}")
````

### A loop inside a loop

Sometimes each item has to be checked against a whole *list* of things: does this caption mention
**any** of several words? That takes a loop inside a loop. The **outer** loop visits each caption, and
the **inner** loop, indented under it, checks that caption against each word. The inner loop runs all
the way through for **each** item of the outer loop.

**Predict first.** Two of these four captions mention a team word. How many items will `matches` hold?

````python
# my guess:
team_words = ["frogs", "tcu", "purple"]
captions = ["Game day!", "Purple everywhere", "Go Frogs, go TCU", "Studying all weekend"]

matches = []
for caption in captions:
    for word in team_words:
        if word in caption.lower():
            matches.append(caption)

print(len(matches), matches)
````

Three items, from two captions. "Go Frogs, go TCU" mentions both *frogs* and *tcu*, so it went in
twice. The `.append()` runs once per **matching word**, and we wanted once per **caption**.

The fix is a **True/False switch**: a variable that remembers whether this caption has matched anything
yet.

The structure:
```python
new_list = []
for item in my_list:
    found = False
    for thing in other_list:
        if thing in item:
            found = True
    if found:
        new_list.append(item)
```

- `found` starts at `False` for **every** item, so it's set inside the outer loop, before the inner one
- The inner loop only flips the switch to `True`. It doesn't append anything
- The `.append()` sits in the **outer** loop, after the inner loop has finished, so each item is added
  once, however many things it matched
- Indentation tells Python which loop each line belongs to, so line it up carefully

````python
# the same captions, with the switch: each caption is added once
matches = []
for caption in captions:
    found = False                  # no match yet for this caption
    for word in team_words:
        if word in caption.lower():
            found = True           # at least one word matched
    if found:                      # back in the outer loop: add the caption once
        matches.append(caption)

print(len(matches), matches)
````

# Putting It All Together

Let's combine everything — lists, methods, loops, and conditionals — to do something that actually looks like digital humanities work.

````python
# analyze hashtag frequency from a mock social media dataset
social_tags = ["#climate", "#action", "#justice", "#climate", "#equality",
               "#justice", "#climate", "#activism", "#equality", "#climate"]

# find unique tags and count each one
analyzed = []

print("Hashtag Frequency Analysis")
print("-" * 30)   # * repeats a string: a line of 30 dashes

for tag in social_tags:
    if tag not in analyzed:
        count = social_tags.count(tag)
        print(f"{tag}: {count} times")
        analyzed.append(tag)

print(f"\nTotal tags: {len(social_tags)}")
print(f"Unique tags: {len(analyzed)}")

# shortcut: set() keeps one copy of each item, so len(set(...)) counts the unique ones
print(f"Unique tags, using set(): {len(set(social_tags))}")
````

# Slice: A Range of Characters

A **slice** grabs a span. The syntax is `text[start : stop]` — it starts at `start` and stops *just
before* `stop`. Leave a side blank to mean "the beginning" or "the end."

```python
print(comment[:9])     # from the start, up to (not including) index 9
print(comment[10:25])  # a middle chunk
print(comment[-13:])   # the last 13 characters
```

how we make them count as the **same word** — a move you'll rely on the moment you start counting
words.

# Replace: Swap One Substring for Another

`.replace(old, new)` hands back a copy with every match swapped. The original string is unchanged —
string methods never edit in place; they **return a new string**. Have a little fun with it first:

```python
print(comment.replace("schools", "group chats"))
print(comment.replace(" ", " 🐸 "))
```

The serious use is **cleaning**. Real text is full of stray newline characters (`\n`) and
junk you want gone before analysis. Replace turns messy text into tidy text.

```python
messy = "God\nand\ncountry"
print(messy)                       # the \n's break it across lines
print("---")
print(messy.replace("\n", " "))   # one clean line
```

# Count & Membership: Asking Questions of the Text

Two quick ways to interrogate a string. `.count()` tells you **how many times** something appears;
the `in` keyword answers **yes or no** — is this substring present?

**Predict first.** Reread the comment: how many times does "god" appear? Is "freedom" in it?

```python
# my guess: "god" appears    times; "freedom" in text ->
text = comment.lower()                 # lowercase first, so case can't fool us
print(text.count("o"))                 # how many letter o's
print(text.count("god"))               # how many times the letters "god" appear
print("commandments" in text)          # is it mentioned at all? -> True / False
print("freedom" in text)
```

Heads up: `.count()` matches **characters**, not whole words — `count("god")` would also catch
the "god" hidden inside *"goddess"* or *"godless."* (Counting real **words** needs `.split()`, coming up
next.) And notice we lowercased **once** at the top and reused it — a habit worth keeping: clean the
text, then ask your questions of the clean version.

# Putting It All Together: Clean, Then Inspect

Here's the real workflow. Take a messy comment, **clean** it (lowercase, drop newlines, trim), then
**inspect** it (split into words, count them, check for a keyword). Every method below you just
learned — chained into one small routine.

```python
raw_comment = "  The Ten Commandments\nDON'T belong in schools.  "

clean = raw_comment.lower().replace("\n", " ").strip()
words = clean.split()

print("Cleaned:", clean)
print("Words:", words)
print("Word count:", len(words))
print("Mentions schools?", "schools" in clean)
```

Read that middle line again: `raw_comment.lower().replace("\n", " ").strip()`. We
**chained** three methods — each returns a new string, which the next one acts on, left to right.
That one line is a preview of how you'll prep real data all term.

**You'll use this, and you'll change it.** Look at what that line erased. `DON'T` in capitals was a
shout, and now it's just `don't`. For **counting words** (HW2), that's exactly what you want: the
shouted and the quiet "don't" should count as one word. For **measuring feeling** (HW3's sentiment
analysis), it's the wrong move, because the sentiment tool reads capitals and `!!!` as emphasis. Same
text, different cleaning, depending on the question you're asking.

# Sneak Preview: Where This Is Going

`.split()` just handed you a **list** of words. On Friday we learn to ask **conditional** questions
(*is this an interesting word, or just "the" and "and"?*), and on Monday to **loop** over lists,
visiting every word one at a time. Put those together and you can do this:

```python
comment = "god god country schools god morals freedom god"
words = comment.split()

counts = {}
for word in words:
    counts[word] = counts.get(word, 0) + 1

print(counts)
```

One new piece: `counts = {}` makes a **dictionary**, not a list. A list keeps items by
position (`words[0]`); a dictionary pairs each word with its count, so you can look one up by name
(`counts["god"]`). Dictionaries get their own session next Wednesday.

Don't worry about every line yet — that's the next three sessions. But notice the shape of it:
**split a text into words, then count them.** That single idea, scaled up to a whole real
comment section, is **term frequency** — the first real analysis you'll run in Week 3, and the start of
hearing what a crowd is actually saying.

It's also the first of the three lenses your capstone can use. **Frequency** asks what a crowd talks
about, **sentiment** asks how it feels, and **topic modeling** asks what sub-conversations it's
having. All three start with a cleaned string, split into words, so the choices you make with today's
methods carry into every one of them.

# The Whole Script

Run this cell. The comment lines in capitals (`# ---- 1. TOOLS ----` and so on) divide it into
six parts.

```python
# ---------- 1. TOOLS: bring in code someone else already wrote ----------
import string            # includes a ready-made string of punctuation marks
import urllib.request    # lets Python download a file from the web

# ---------- 2. FUNCTIONS: name a routine once, reuse it anywhere ----------
def split_into_words(text):
    """Lowercase the text, split it on spaces, and trim punctuation off each word."""
    words = []
    for word in text.lower().split():
        word = word.strip(string.punctuation + "“”‘’")
        if word != "":
            words.append(word)
    return words

def count_words(word_list):
    """Return a dictionary: each word -> how many times it appears."""
    counts = {}
    for word in word_list:
        if word in counts:
            counts[word] = counts[word] + 1
        else:
            counts[word] = 1
    return counts

# ---------- 3. SETTINGS: the choices we make before counting ----------
URL = ("https://raw.githubusercontent.com/TCU-DCDA/WRIT20833_2026/main/"
       "notebooks/data/discussions/d1_discussion_posts.txt")
HOW_MANY = 20            # how many top words to show
STOPWORDS = ["the", "i", "to", "a", "and", "in", "of", "that", "who", "on", "for", "was",
             "it", "or", "as", "is", "my", "be", "me", "this", "with", "but", "not", "what",
             "are", "by", "how", "an", "at", "one", "can", "have", "so", "they", "would",
             "because", "you", "your", "from", "people", "more", "like", "also", "just",
             "their", "them", "which", "other", "being", "into", "could", "these", "when",
             "were", "out", "there", "has", "had", "do", "all", "some", "about", "if", "we",
             "our", "us", "he", "she", "his", "her", "been", "than", "then", "its", "it's",
             "i'm", "don't", "very", "even", "only", "way", "much", "many", "think"]

# ---------- 4. READ: load the file ----------
text = urllib.request.urlopen(URL).read().decode("utf-8")

# ---------- 5. ANALYZE: split, filter, count, sort ----------
all_words = split_into_words(text)

kept_words = []
for word in all_words:
    if word not in STOPWORDS:
        kept_words.append(word)

counts = count_words(kept_words)
top = sorted(counts.items(), key=lambda item: item[1], reverse=True)[:HOW_MANY]

# ---------- 6. OUTPUT: show the results ----------
print(len(all_words), "words in all;", len(kept_words), "left after removing stopwords\n")
for word, n in top:
    print(n, word)
```

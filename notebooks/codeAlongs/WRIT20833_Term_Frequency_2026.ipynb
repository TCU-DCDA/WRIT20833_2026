# Setup

Run this cell. It hands you three tools. **Two are new today:** a `split_into_words` helper (like
`.split()`, but it also strips punctuation) and a long `stopwords` skip-list (the full version of the
five-word preview in HW1 Part A). **One you've seen:** Python's built-in `Counter`, from the end of the
Dictionaries & Functions code-along. You'll *use* them today; you don't need to read every line yet.

A reminder of how `Counter` works:
```python
from collections import Counter     # load a tool that comes with Python

Counter(my_list)                    # counts each item, like count_words
Counter(my_list).most_common(n)     # the n most frequent items, as (item, count) pairs
```

````python
import re
from collections import Counter

def split_into_words(any_chunk_of_text):
    lowercase_text = any_chunk_of_text.lower()
    # split on any run of non-letter/number characters (spaces, punctuation, newlines)
    return re.split(r"\W+", lowercase_text)

stopwords = ["i", "me", "my", "myself", "we", "our", "ours", "ourselves", "you", "your", "yours", "yourself", "yourselves", "he", "him", "his", "himself", "she", "her", "hers", "herself", "it", "its", "itself", "they", "them", "their", "theirs", "themselves", "what", "which", "who", "whom", "this", "that", "these", "those", "am", "is", "are", "was", "were", "be", "been", "being", "have", "has", "had", "having", "do", "does", "did", "doing", "a", "an", "the", "and", "but", "if", "or", "because", "as", "until", "while", "of", "at", "by", "for", "with", "about", "against", "between", "into", "through", "during", "before", "after", "above", "below", "to", "from", "up", "down", "in", "out", "on", "off", "over", "under", "again", "further", "then", "once", "here", "there", "when", "where", "why", "how", "all", "any", "both", "each", "few", "more", "most", "other", "some", "such", "no", "nor", "not", "only", "own", "same", "so", "than", "too", "very", "s", "t", "can", "will", "just", "don", "should", "now", "ve", "ll", "amp"]

print("Ready. The stopwords skip-list has", len(stopwords), "words in it.")
````

### What `split_into_words` does

The helper's one working line is `re.split(r"\W+", lowercase_text)`. That `r"\W+"` is a **regular
expression** (a "regex" for short): a tiny language for describing patterns in text. You won't need to
write one in this course. This one means *any run of characters that aren't a letter, a number, or an
underscore*, and the helper splits the text wherever it finds one. Spaces, punctuation and line breaks
all count, and so does the apostrophe.

**Predict first.** `.split()` from the Strings code-along splits only at spaces. How will the two
results differ for this sentence?

````python
# my guess: the two lists will differ in
sentence = "Don't put God's law in schools!"

print(sentence.lower().split())
print(split_into_words(sentence))
````

`.split()` kept `schools!` with its `!`, so it would count as a different word from `schools`.
`split_into_words` fixes that, but it also breaks `don't` into `don` and `t`, and `god's` into `god`
and `s`. That's why the stopword list includes `"don"`, `"t"`, `"s"`, `"ve"` and `"ll"`: someone noticed
the leftover pieces and decided to throw them out. The empty string `''` at the end is what's left after
the final `!`. Every tool that cuts text into words makes choices like these, and the counts inherit
them.

# Part 1 — Count Every Word

Here's a small pile of real-sounding comments about the Texas law that puts the Ten Commandments in
classrooms. Let's count what's in it. The code below is a pattern you know from Lists & Loops
("Building new lists from old ones"): start with an empty list, loop over the words, and `.append()`
each one that passes a test. Then `Counter` tallies the list and hands back the **10 most common**.

One detail: like the `!` above, the line breaks at the start and end of this text leave an empty
string `""` at each end. The `if` skips those.

**Predict first.** Which single word will come out on top? Write your guess in the comment, then run.

````python
# my guess: the most common word will be
comments_text = """
The Ten Commandments belong in every classroom in this country, and our kids need them.
Morals matter, and our schools are where our kids learn about right and wrong.
God and country are what built this nation, and the schools should teach that.
It is common sense to put these values back in the schools.
Religion has no place in a public school that is paid for by all of us.
And whose religion are we talking about? Religion is for the home, not the schools.
Freedom of religion has to mean freedom from religion in our public schools too.
The government should not be putting religion or god into the schools of this country.
"""

all_words = []
for word in split_into_words(comments_text):
    if word != "":                  # skip the empty strings at the edges
        all_words.append(word)

print("Total words:", len(all_words))
print(Counter(all_words).most_common(10))
````

Look at the very top of that list: **the** and **and** — the little connective words that hold
*any* English sentence together, carrying nothing about the topic. (A couple of content words like
*schools* and *religion* are climbing up too, because they repeat a lot — but they're elbow to elbow with
noise.) The single most frequent words are meaningless, which makes the whole ranking untrustworthy.
Let's clear the noise out so the count actually means something.

# Part 3 — Make It Reusable

You'll want to run this on text after text, so wrap it in a **function**, as in the Dictionaries &
Functions code-along. Give it a text and a number `n`; get back the top `n` meaningful words. Inside,
it's the Part 2 loop indented one level under `def`, with `text` standing in for `comments_text`.
Write the recipe once, reuse it forever.

````python
def top_meaningful_words(text, n):
    meaningful_words = []
    for word in split_into_words(text):
        if word != "" and word not in stopwords:
            meaningful_words.append(word)
    return Counter(meaningful_words).most_common(n)

print(top_meaningful_words(comments_text, 5))
````

# Part 4 — Whose Words Win?

Term frequency gets *interesting* when you **compare two texts**. Here's a second one, in the flat,
official voice of a founding legal document — the kind of thing those commenters keep invoking. It is a
**constructed paraphrase**: lines from the Constitution and its later amendments, shortened and reworded
for this exercise, so don't quote it as the Constitution itself. (HW2 uses the real 1787 text.)
Run the same function on both and watch two completely different vocabularies appear.

````python
# type along — run top_meaningful_words on BOTH texts (top 5 each) and print them side by side, labeled
official_text = """
Congress shall pass no law respecting an establishment of religion.
The powers not delegated to the United States by the Constitution are reserved to the States.
All legislative powers shall be vested in a Congress of the United States.
No State shall make any law which shall abridge the privileges of citizens.
The Congress shall have power to make all laws necessary and proper.
"""
````

Same method, two different *aboutnesses*. The official text runs on **shall, congress, states,
law**; the comments on **schools, religion, country, kids**. Neither list is the "right" one —
together they show two voices talking about the same issue in completely different words. That gap *is*
the analysis.

We can even name the words that are distinctive to one side — the comments' top words that the
official text never uses **at all**. That means checking against *every* word in the official text, not
just its top few (two top-8 lists can easily share nothing, which would make every word look
"distinctive").

Two pieces make this work. `set(...)` keeps one copy of each word, which is all we need to ask "is this
word anywhere in the text?" And each item `top_meaningful_words` hands back is a **pair**, a word and its
count, so the loop can name both halves at once:

The structure:
```python
for first_part, second_part in list_of_pairs:
    do something with first_part and second_part
```

- Each pair is split in two as the loop visits it, the same way `enumerate()` gave you a position
  and an item in Lists & Loops
- Name the two parts for what they hold, here `word` and `count`

````python
official_words = set(split_into_words(official_text))   # every word the official text uses

for word, count in top_meaningful_words(comments_text, 8):
    if word in official_words:
        print(f"shared       {word} ({count})")
    else:
        print(f"distinctive  {word} ({count})")
````

Almost everything the commenters care about — **schools, kids, god, freedom** — is a word the legal
text never uses. The one word they share is **religion**. That is where the two voices actually meet,
and it's the word they are arguing over.

# Putting It All Together

The whole term-frequency move, start to finish — split, drop the stopwords, count, take the top. This is
the engine you'll run on real data in HW2.

````python
def top_meaningful_words(text, n):
    meaningful_words = []
    for word in split_into_words(text):
        if word != "" and word not in stopwords:
            meaningful_words.append(word)
    return Counter(meaningful_words).most_common(n)

print("Comments ->", top_meaningful_words(comments_text, 5))
print("Official ->", top_meaningful_words(official_text, 5))
````

# Step 1 — The loop you already know

Say we want the **number of words in each comment**. Here is the Day 3–4 way: start with an empty
list, loop, and `.append()` one result per item.

````python
word_counts = []
for c in comments:
    word_counts.append(len(c.split()))

print(word_counts)
````

**Check:** `[7, 8, 7, 6, 6, 4, 8, 12]`

Now the same list, built in one line:

````python
word_counts_short = [len(c.split()) for c in comments]

print(word_counts_short)
print(word_counts == word_counts_short)
````

Same list, and the last line prints `True`. Put the two side by side and the pieces line up:

```
word_counts = []                              word_counts_short = [ len(c.split())   for c in comments ]
for c in comments:                                                    ^ WHAT to keep   ^ the loop line
    word_counts.append( len(c.split()) )
```

**The pattern:** `[ WHAT-TO-KEEP   for ITEM in LIST ]`

Read it out loud as a sentence: *"make a list of `len(c.split())`, for each `c` in `comments`."*

# Step 3 — Adding a filter

The loops you wrote in Term Frequency had an `if` inside: *keep only the items that pass a test.* A
comprehension can do that too. Here is the loop version first: keep only the comments that mention
**religion**.

````python
about_religion = []
for c in comments:
    if "religion" in c.lower():
        about_religion.append(c)

print(len(about_religion), "comments")
for c in about_religion:
    print("-", c)
````

**Check:** `3 comments`, then the three comments that contain *religion*.

# What to take away

- A list comprehension is a loop that builds a list, on one line:
  `[ WHAT-TO-KEEP  for ITEM in LIST  if TEST ]` (the `if TEST` part is optional).
- **To read one,** turn it into a loop: empty list, `for`, `if`, `.append()`.
- **To write one,** you can always write the loop first.
- Neither form is "the right one." Choose the version you can explain to someone else.

When you meet a `[ ... for ... in ... ]` in the Topic Modeling notebook, in homework starter code, or
in something an AI wrote, you now know how to read it.
